# Projectile Motion Simulator - Implementation

## Goals - Phase 1:
   1. Simulate motion using kinematic equations
   2. Display motion using vectors or an animation
   3. Accept user input to change motion variables

## Goals - Phase 2:
   1. Implement drag coefficients
   2. Compare ideal case vs. case with drag

## Equations:
   - Free-Fall Acceleration: 1/2gt^2

   - Vector Components:
     - Vx0 = V0 * cos(theta)
     - Vy0 = V0 * sin(theta)
     - V = sqrt(vx^2 + vy^2)

   - Kinematic Equations (no drag):
     - X(t) = X0 + Vx0t <- no drag
     - Y(t) = Y0 + Vy0t - 1/2gt^2
     - Vx(t) = Vx0 <- no drag
     - Vy(t) = Vy0 - gt

### Adding Drag:
    To add drag we will need to convert the kinematic equations from linear into rates of change using calculus
    for example to find the x-component for velocity, the change in x over the change in time = vx (dx/dt = vx)

    We will have to use Euler's method to perform numerical integration for each timestep

```python
g = 9.81

def calculate_magnitude(vector):
    magnitude = np.linalg.norm(vector)
    # V = sqrt(Vx^2 + Vy^2)
    # np method for: magnitude = (vector[0]**2 + vector[1]**2) ** 0.5
    return magnitude

def calculate_pos(pos0, vec0, t):
    xpos = pos0[0] + (vec0[0] * t)
    # X(t) = X0 + Vx0t
    ypos = pos0[1] + (vec0[1] * t) - (0.5*g*t**2)
    # Y(t) = Y0 + VY0t - 1/2gt^2
    # Can also just write as pos = pos0 + vec0 since they are vectors
    return np.array([xpos, ypos])

def calculate_comp(mag, theta):
    vx = mag * np.cos(theta)
    # Vx0 = V0 * cos(theta)
    vy = mag * np.sin(theta)
    # Vy0 = V0 * sin(theta)
    return np.array([vx, vy])

def calculate_vx(vector0, t=0):
    vx = vector0[0]
    # Vx(t) = Vx0 <-- no drag
    vy = vector0[1] - (g * t)
    # Vy(t) = Vy0 - gt
    return np.array([vx, vy])
```

```python
vector1 = np.array([30*(np.cos(np.radians(45))), 30*(np.sin(np.radians(45)))])
print(f"Vector 1 Initial: {vector1}")

position1 = np.array([0, 15])
print(f"Position 1 Initial: {position1}")

t = 4.8
print(f"Time: {t}")

mag = calculate_magnitude(vector1)
print(f"Magnitude: {mag}")

pos = calculate_pos(position1, vector1, t)
print(f"X position after {t} seconds: {pos[0]} ")
print(f"Y position after {t} seconds: {pos[1]} ")


position_history = [position1]
velocity_history = [vector1]
time_history = [0]

t = 0
dt = 1

for i in range(6):
    t += dt
    position1 = calculate_pos(position1, vector1, dt)
    vector1 = calculate_vx(vector1, dt)

    position_history.append(position1)
    velocity_history.append(vector1)
    time_history.append(t)


positions = np.array(position_history)
velocities = np.array(velocity_history)
time_array = np.array(time_history)
speeds = np.linalg.norm(velocities, axis=1)

plt.plot(time_array, positions[:, 1])
plt.xlabel("Time (s)")
plt.ylabel("Height (m)")
plt.show()

plt.plot(time_array, speeds)
plt.xlabel("Time (s)")
plt.ylabel("Speed (m/s)")
plt.show()

plt.plot(positions[:, 0], positions[:, 1])
plt.xlabel("Distance (m)")
plt.ylabel("Height (m)")
plt.show()

plt.plot(time_array, velocities[:, 1])
plt.xlabel("Time (s)")
plt.ylabel("Vy (m/s)")
plt.axhline(0, color='red', linestyle='--')
```
